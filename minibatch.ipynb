```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
```

```python
X, Y= make_blobs(n_samples=1500,centers=3, n_features=5, random_state=42)
```

```python
mbk_means = MiniBatchKMeans(n_clusters=5, batch_size=200,random_state=42)
```

```python
mbk_means.fit(X)
```

```python
plt.scatter(X[:,0],X[:,1], c=mbk_means.labels_,cmap='viridis')
plt.scatter(mbk_means.cluster_centers_[:, 0], mbk_means.cluster_centers_[:, 1],s=200,c='red',marker='X')
```
